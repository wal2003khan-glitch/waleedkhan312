```python
#calculation function
def shop(price):
  print(f"price is:{price}")
  def add_tax():
    tax = price*0.10
    print(f"tax:{tax}")
    return price+tax
  total=add_tax()
  print(f"total:{total}")
shop(50)
```

```python
#function calculation
def shop(price):
  print("price is ",price)
  def add_tax():
    tax=price+0.1
    print("tax is ",tax)
    return price+tax
  total=add_tax()
  print("total is ",total)
shop(30)
```

Example 15

```python
#factorial function
def factorial_recursive(n):
  if n<=1:
    return 1
  return n*factorial_recursive(n-1)
print(factorial_recursive(5))

```
